## Cleaning data

```python
cols_to_convert = ["met", "xmet", "smet"]
df[cols_to_convert] = df[cols_to_convert].replace('*', 0).apply(pd.to_numeric)
```

## Exploring jobs

```python
df_jobs = df[['Nom métier BMO', 'Lbl_fam_met']]
df_unique = df_jobs.drop_duplicates(subset='Nom métier BMO')
df_sorted = df_unique.sort_values(by='Lbl_fam_met')
grouped = df_sorted.groupby('Lbl_fam_met')
```

```python
for fam, group in grouped:
    print(f"\n=== {fam} ===")
    for job in group['Nom métier BMO']:
        print(f"- {job}")
```
